## Practice problem: RemoteOK job listings


### Task

Collect structured job listings from the RemoteOK public API (`https://remoteok.com/api`).

Fields:

- Company name
- Job role
- Location
- Tags (technologies, benefits, job type)

RemoteOK exposes JSON. Use the API, not HTML scraping.

Output: `remoteok_jobs.csv`.


```python
import requests
import pandas as pd

STUDENT = {
    "name": "Francis Reuben R",
    "usn": "1RUA25CSE0168",
    "section": "G",
}
HEADERS = {
    "User-Agent": "DS-Lab-Exp1/1.0 (Francis Reuben R; USN 1RUA25CSE0168; Section G; academic)"
}

api_url = "https://remoteok.com/api"
print(STUDENT)
print("Fetching", api_url)

response = requests.get(api_url, headers=HEADERS, timeout=30)
response.raise_for_status()
payload = response.json()

# First element is metadata; job objects follow.
job_listings = [item for item in payload if isinstance(item, dict) and "position" in item]

rows = []
for job in job_listings:
    tags = job.get("tags") or []
    rows.append({
        "Company Name": job.get("company", "N/A"),
        "Job Role": job.get("position", "N/A"),
        "Location": job.get("location") or "Worldwide",
        "Tags": ", ".join(tags) if isinstance(tags, list) else str(tags),
    })

jobs_df = pd.DataFrame(rows)
print("Collected", len(jobs_df), "job listings.")
jobs_df.head()
```

```python
out_path = "remoteok_jobs.csv"
if jobs_df.empty:
    print("No rows collected. CSV not written.")
else:
    jobs_df.to_csv(out_path, index=False, encoding="utf-8")
    print("Saved", out_path)
    print(jobs_df["Location"].value_counts().head(10))
```

### Result

Job listings are stored in `remoteok_jobs.csv` with company, role, location, and tags.

### Conclusion

The RemoteOK JSON API was queried with `requests`, parsed into a DataFrame, and written to CSV. This is the same collect-then-store pattern used for Open-Meteo and the bookstore scrape.
