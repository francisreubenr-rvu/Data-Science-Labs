## Quick Reference

| | Z-test | t-test |
|---|---|---|
| Use when | Population σ is **known** | Population σ is **unknown** |
| Formula | $z = \dfrac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$ | $t = \dfrac{\bar{x} - \mu_0}{s / \sqrt{n}}$, $df = n - 1$ |

| Tail | $H_1$ | p-value (Z) | p-value (t) | Critical value (Z, α = 0.05) |
|---|---|---|---|---|
| Left | $\mu < \mu_0$ | `stats.norm.cdf(z)` | `stats.t.cdf(t, dof)` | −1.645 |
| Right | $\mu > \mu_0$ | `1 - stats.norm.cdf(z)` | `1 - stats.t.cdf(t, dof)` | +1.645 |
| Two | $\mu \neq \mu_0$ | `2 * (1 - stats.norm.cdf(abs(z)))` | `2 * (1 - stats.t.cdf(abs(t), dof))` | ±1.960 |

**Decision rule:** if **p-value ≤ α**, reject $H_0$. Otherwise, fail to reject $H_0$.

For the sample standard deviation use `df['column'].std()`. pandas uses `ddof=1` by default, which is what the t-test needs.

⚠️ The dataset is stored in a variable called `df`. Store the degrees of freedom in a variable called **`dof`**, not `df`, or you will overwrite the dataset.

# Part A – Z-Tests (σ known, use all 40 students)

## Q1. Z-Test – Left-Tailed

A health report says college students sleep **7 hours** per night on average, with a known population standard deviation of **σ = 1.2 hours**.
Using `sleep_hours` for **all 40 students**, test at α = 0.05 whether students sleep **less than 7 hours** on average.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# Q1 – Z-test, left-tailed: sleep_hours, mu0 = 7, sigma = 1.2, n = 40
mu0 = 7
sigma = 1.2
data = df['sleep_hours']
n = len(data)
xbar = data.mean()

z = (xbar - mu0) / (sigma / np.sqrt(n))
p_value = stats.norm.cdf(z)
crit = stats.norm.ppf(alpha)

print("H0: mu = 7")
print("H1: mu < 7")
print("Tail: left")
print(f"n = {n}")
print(f"xbar = {xbar:.4f}")
print(f"z = {z:.4f}")
print(f"critical value = {crit:.4f}")
print(f"p-value = {p_value:.4f}")
if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

**Your answer – Q1**

- **$H_0$:** $\mu = 7$ (students sleep 7 hours per night on average)
- **$H_1$:** $\mu < 7$ (students sleep less than 7 hours)
- **Tail (left / right / two):** left
- **Test statistic:** $z = -1.8578$  ($\bar{x} = 6.6475$, $n = 40$, $\sigma = 1.2$)
- **Critical value:** $-1.6449$
- **p-value:** $0.0316$
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Reject $H_0$
- **Conclusion in words:** At $\alpha = 0.05$, there is enough evidence that college students sleep less than 7 hours per night on average.


## Q2. Z-Test – Right-Tailed

A national survey says young adults spend **5 hours** a day on screens, with a known population standard deviation of **σ = 1.5 hours**.
Using `screen_time` for **all 40 students**, test at α = 0.05 whether these students' screen time is **more than 5 hours** a day.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# Q2 – Z-test, right-tailed: screen_time, mu0 = 5, sigma = 1.5, n = 40
mu0 = 5
sigma = 1.5
data = df['screen_time']
n = len(data)
xbar = data.mean()

z = (xbar - mu0) / (sigma / np.sqrt(n))
p_value = 1 - stats.norm.cdf(z)
crit = stats.norm.ppf(1 - alpha)

print("H0: mu = 5")
print("H1: mu > 5")
print("Tail: right")
print(f"n = {n}")
print(f"xbar = {xbar:.4f}")
print(f"z = {z:.4f}")
print(f"critical value = {crit:.4f}")
print(f"p-value = {p_value:.4f}")
if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

**Your answer – Q2**

- **$H_0$:** $\mu = 5$ (mean daily screen time is 5 hours)
- **$H_1$:** $\mu > 5$ (mean daily screen time is more than 5 hours)
- **Tail (left / right / two):** right
- **Test statistic:** $z = 2.0028$  ($\bar{x} = 5.4750$, $n = 40$, $\sigma = 1.5$)
- **Critical value:** $+1.6449$
- **p-value:** $0.0226$
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Reject $H_0$
- **Conclusion in words:** At $\alpha = 0.05$, these students spend more than 5 hours a day on screens.


## Q3. Z-Test – Two-Tailed

Over many years, the course's final exam has had a mean of **65** with a known standard deviation of **σ = 10**.
Using `exam_score` for **all 40 students**, test at α = 0.05 whether this year's mean score is **different from 65**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# Q3 – Z-test, two-tailed: exam_score, mu0 = 65, sigma = 10, n = 40
mu0 = 65
sigma = 10
data = df['exam_score']
n = len(data)
xbar = data.mean()

z = (xbar - mu0) / (sigma / np.sqrt(n))
p_value = 2 * (1 - stats.norm.cdf(abs(z)))
crit = stats.norm.ppf(1 - alpha / 2)

print("H0: mu = 65")
print("H1: mu != 65")
print("Tail: two")
print(f"n = {n}")
print(f"xbar = {xbar:.4f}")
print(f"z = {z:.4f}")
print(f"critical values = ±{crit:.4f}")
print(f"p-value = {p_value:.4f}")
if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

**Your answer – Q3**

- **$H_0$:** $\mu = 65$ (this year's mean exam score is 65)
- **$H_1$:** $\mu \neq 65$ (this year's mean exam score is different from 65)
- **Tail (left / right / two):** two
- **Test statistic:** $z = -0.5060$  ($\bar{x} = 64.2000$, $n = 40$, $\sigma = 10$)
- **Critical value:** $\pm 1.9600$
- **p-value:** $0.6129$
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Fail to reject $H_0$
- **Conclusion in words:** At $\alpha = 0.05$, there is not enough evidence that this year's mean exam score differs from 65.

## Q4. t-Test – Left-Tailed

The class teacher of **Section B** believes her students' attendance is **below 83 %**. The population standard deviation is **not known**.
Using `attendance` for **Section B only**, test at α = 0.05 whether the mean attendance is **less than 83 %**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# Q4 – t-test, left-tailed: attendance, Section B only, mu0 = 83
mu0 = 83
data = df[df['section'] == 'B']['attendance']
n = len(data)
xbar = data.mean()
s = data.std()          # pandas default ddof=1
dof = n - 1

t = (xbar - mu0) / (s / np.sqrt(n))
p_value = stats.t.cdf(t, dof)
crit = stats.t.ppf(alpha, dof)

print("H0: mu = 83")
print("H1: mu < 83")
print("Tail: left")
print(f"n = {n}")
print(f"xbar = {xbar:.4f}")
print(f"s = {s:.4f}")
print(f"dof = {dof}")
print(f"t = {t:.4f}")
print(f"critical value = {crit:.4f}")
print(f"p-value = {p_value:.4f}")
if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")

result = stats.ttest_1samp(data, mu0, alternative='less')
print("SciPy check:", result)
```

**Your answer – Q4**

- **$H_0$:** $\mu = 83$ (Section B mean attendance is 83%)
- **$H_1$:** $\mu < 83$ (Section B mean attendance is less than 83%)
- **Tail (left / right / two):** left
- **Test statistic:** $t = -1.0307$  ($\bar{x} = 80.6154$, $s = 8.3420$, $n = 13$, $df = 12$)
- **Critical value:** $-1.7823$
- **p-value:** $0.1615$
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Fail to reject $H_0$
- **Conclusion in words:** At $\alpha = 0.05$, there is not enough evidence that Section B attendance is below 83%. SciPy `ttest_1samp(..., alternative='less')` matches the hand calculation.


## Q5. t-Test – Right-Tailed

The college recommends at least **2.5 hours** of self-study per day. The teacher of **Section A** claims her students study **more than 2.5 hours** a day on average. σ is **not known**.
Using `study_hours` for **Section A only**, test the claim at α = 0.05.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# Q5 – t-test, right-tailed: study_hours, Section A only, mu0 = 2.5
mu0 = 2.5
data = df[df['section'] == 'A']['study_hours']
n = len(data)
xbar = data.mean()
s = data.std()
dof = n - 1

t = (xbar - mu0) / (s / np.sqrt(n))
p_value = 1 - stats.t.cdf(t, dof)
crit = stats.t.ppf(1 - alpha, dof)

print("H0: mu = 2.5")
print("H1: mu > 2.5")
print("Tail: right")
print(f"n = {n}")
print(f"xbar = {xbar:.4f}")
print(f"s = {s:.4f}")
print(f"dof = {dof}")
print(f"t = {t:.4f}")
print(f"critical value = {crit:.4f}")
print(f"p-value = {p_value:.4f}")
if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")

result = stats.ttest_1samp(data, mu0, alternative='greater')
print("SciPy check:", result)
```

**Your answer – Q5**

- **$H_0$:** $\mu = 2.5$ (Section A mean self-study is 2.5 hours per day)
- **$H_1$:** $\mu > 2.5$ (Section A students study more than 2.5 hours per day)
- **Tail (left / right / two):** right
- **Test statistic:** $t = 2.1185$  ($\bar{x} = 2.9286$, $s = 0.7569$, $n = 14$, $df = 13$)
- **Critical value:** $+1.7709$
- **p-value:** $0.0270$
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Reject $H_0$
- **Conclusion in words:** At $\alpha = 0.05$, Section A students study more than 2.5 hours a day on average. SciPy `ttest_1samp(..., alternative='greater')` matches the hand calculation.


## Q6. t-Test – Two-Tailed

The transport office plans the college bus timetable assuming an average one-way commute of **35 minutes**. σ is **not known**.
Using `commute_min` for **Section C only**, test at α = 0.05 whether the mean commute time is **different from 35 minutes**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# Q6 – t-test, two-tailed: commute_min, Section C only, mu0 = 35
mu0 = 35
data = df[df['section'] == 'C']['commute_min']
n = len(data)
xbar = data.mean()
s = data.std()
dof = n - 1

t = (xbar - mu0) / (s / np.sqrt(n))
p_value = 2 * (1 - stats.t.cdf(abs(t), dof))
crit = stats.t.ppf(1 - alpha / 2, dof)

print("H0: mu = 35")
print("H1: mu != 35")
print("Tail: two")
print(f"n = {n}")
print(f"xbar = {xbar:.4f}")
print(f"s = {s:.4f}")
print(f"dof = {dof}")
print(f"t = {t:.4f}")
print(f"critical values = ±{crit:.4f}")
print(f"p-value = {p_value:.4f}")
if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")

result = stats.ttest_1samp(data, mu0, alternative='two-sided')
print("SciPy check:", result)
```

**Your answer – Q6**

- **$H_0$:** $\mu = 35$ (Section C mean one-way commute is 35 minutes)
- **$H_1$:** $\mu \neq 35$ (Section C mean commute is different from 35 minutes)
- **Tail (left / right / two):** two
- **Test statistic:** $t = -2.8611$  ($\bar{x} = 29.2308$, $s = 7.2704$, $n = 13$, $df = 12$)
- **Critical value:** $\pm 2.1788$
- **p-value:** $0.0143$
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Reject $H_0$
- **Conclusion in words:** At $\alpha = 0.05$, Section C mean commute time is different from 35 minutes (the sample mean is lower). SciPy `ttest_1samp(..., alternative='two-sided')` matches the hand calculation.
